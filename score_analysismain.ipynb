```python
import pandas as pd

# Sample cricket match data
data = {
    "Player": ["Rohit Sharma", "KL Rahul", "Virat Kohli", "Suryakumar Yadav", "Hardik Pandya",
               "MS Dhoni", "Ravindra Jadeja", "Bhuvneshwar Kumar", "Jasprit Bumrah", "Yuzvendra Chahal", "Mohammed Shami"],
    "Runs": [45, 67, 38, 75, 25, 50, 15, 18, 8, 5, 8],
    "Wickets": [0, 0, 0, 0, 1, 0, 2, 3, 1, 2, 4],
    "Balls Faced": [30, 50, 42, 55, 20, 45, 12, 8, 2, 4, 11],
    "Balls Bowled": [0, 0, 0, 0, 24, 6, 60, 60, 48, 48, 54],
    "Runs Conceded": [0, 0, 0, 0, 35, 12, 50, 49, 45, 60, 40],
    "Dot Balls": [0, 0, 0, 0, 6, 0, 15, 10, 5, 7, 12]
}

# Create DataFrame
match_data = pd.DataFrame(data)

# Extras
extras = {
    "Wides": 5,
    "Byes": 3,
    "No Balls": 2,
    "Leg Byes": 4
}
total_extras = sum(extras.values())

# Total team runs (including extras)
total_runs = match_data['Runs'].sum()
total_runs_with_extras = total_runs + total_extras
print("Total Runs (excluding extras):", total_runs)
print("Extras:", extras)
print("Total Runs (including extras):", total_runs_with_extras)

# Calculate Strike Rate for each player
match_data['Strike Rate'] = (match_data['Runs'] / match_data['Balls Faced']) * 100

# Filter out bowlers who have bowled at least one ball
bowlers = match_data[match_data['Balls Bowled'] > 0].copy()

# Calculate Bowling Strike Rate (Balls Bowled per Wicket)
bowlers['Bowling Strike Rate'] = bowlers['Balls Bowled'] / bowlers['Wickets']

# Calculate Economy Rate (Runs Conceded per Over Bowled)
bowlers['Economy Rate'] = bowlers['Runs Conceded'] / (bowlers['Balls Bowled'] / 6)

# Calculate Dot Ball Percentage for each bowler
bowlers['Dot Ball Percentage'] = (bowlers['Dot Balls'] / bowlers['Balls Bowled']) * 100

# Points system: 1 point per 10 runs, 20 points per wicket
match_data['Points'] = (match_data['Runs'] // 10) + (match_data['Wickets'] * 20)

# Find key players based on overall impact
# Sort players by points
key_players = match_data[['Player', 'Runs', 'Wickets', 'Points', 'Strike Rate']].sort_values(by='Points', ascending=False)

# Best all-rounder (players with good performance in both batting and bowling)
all_rounders = match_data[(match_data['Runs'] > 20) & (match_data['Wickets'] > 0)].sort_values(by='Points', ascending=False)

# Best Batsman based on runs and strike rate
best_batsman = match_data.loc[match_data['Runs'].idxmax()]

# Best Bowler based on wickets, economy rate, and dot ball percentage
best_bowler = bowlers.loc[bowlers['Wickets'].idxmax()]

# **Key Turning Point**: Identify match-changing contributions
# Here, let's consider the best batting and bowling performance in the match as a potential turning point
# Turning point based on top batsman and bowler contribution (in terms of runs and wickets)

turning_point_batsman = best_batsman['Player']
turning_point_batsman_runs = best_batsman['Runs']

turning_point_bowler = best_bowler['Player']
turning_point_bowler_wickets = best_bowler['Wickets']

# Display Key Player Analysis and Turning Point
print("\nKey Players Analysis (Top by Points):")
print(key_players[['Player', 'Runs', 'Wickets', 'Points', 'Strike Rate']])

print("\nBest Batsman based on Runs and Strike Rate:")
print(best_batsman[['Player', 'Runs', 'Strike Rate']])

print("\nBest Bowler based on Wickets, Economy Rate, and Dot Balls:")
print(best_bowler[['Player', 'Wickets', 'Economy Rate', 'Dot Ball Percentage']])

print("\nTop All-rounders:")
print(all_rounders[['Player', 'Runs', 'Wickets', 'Points']])

# **Key Turning Point Information**
print("\nKey Turning Point (Best Player Contributions):")
print(f"Batsman with the highest impact: {turning_point_batsman} with {turning_point_batsman_runs} runs")
print(f"Bowler with the highest impact: {turning_point_bowler} with {turning_point_bowler_wickets} wickets")

```
